```python
import pandas as pd
import numpy as np
data = {
'Order_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'Customer_Name': ['Amara', 'Bola', None, 'Diana', 'Efua', 'Fatima', 'Grace',
None, 'Ifeoma', 'Joyce'],
'Country': ['Ghana', 'Nigeria', 'Kenya', None, 'Ghana', 'Cameroon', None,
'Kenya', 'Nigeria', 'Ghana'],
'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', None, 'Phone', 'Tablet',
'Laptop', None, 'Phone'],
'Quantity': [2, 1, None, 3, 2, None, 1, 4, 2, None],
'Price': [1500, 800, 600, 1500, None, 800, 600, 1500, None, 800],
'Delivery_Days': [3, None, 7, 5, 3, None, 6, None, 4, 5]
}
df = pd.DataFrame(data)
print(df)
```

```python
#. TASK 1
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

TASK 1 - EXPLANATION : 
The dataset contains missing values in Customer_Name, Country, Product, Quantity, Price, and Delivery_Days.
Quantity and Delivery_Days have the highest number of missing values.
The data needs to be cleaned before analysis can be performed.

```python
#TASK 2
df["Customer_Name"] = df["Customer_Name"].fillna("Unknown")

df
```

TASK 2- EXPANATION :
I filled the missing customer names with "Unknown" because the actual names cannot be recovered, but the orders are still useful and should remain in the dataset.

```python
#. TASK 3
print("Mean Quantity:", df["Quantity"].mean())
print("Mean Price:", df["Price"].mean())
```

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())

df["Price"] = df["Price"].fillna(df["Price"].mean())

df
```

TASK 3- EXPLANATION :
For Quantity, I used the mean because the values are fairly close together and there are no extreme values.

For Price, I used the mean because the prices are reasonably balanced and there are no significant outliers.

```python
#. TASK 4
df["Delivery_Days"].describe()
```

The delivery days range from 3 to 7 days, so there are no obvious outliers. Since the values are close together, the mean is an appropriate choice for filling the missing values.

```python
df["Delivery_Days"] = df["Delivery_Days"].fillna(df["Delivery_Days"].mean())

df
```

```python
#. TASK 5
print("Country Mode:", df["Country"].mode()[0])
print("Product Mode:", df["Product"].mode()[0])
```

```python
df["Country"] = df["Country"].fillna(df["Country"].mode()[0])

df["Product"] = df["Product"].fillna(df["Product"].mode()[0])

df
```

TASK 5- EXPLANATION : 
I filled the missing Country values using the mode because Country is a categorical column and the most common value is a reasonable replacement.

I filled the missing Product values using the mode because Product is also a categorical column and using the most frequent product helps preserve the dataset.

```python
#. TASK 6
df.isnull().sum()
```

TASK 6- EXPLANATION : 
The output shows that all columns now contain zero missing values, which means the dataset has been successfully cleaned.

#. TASK 7 - REFLECTION :
The hardest decision was choosing how to handle the Country and Product columns because there are different ways to deal with missing categorical data. I did not use the same method for every column because different data types require different cleaning techniques. For Customer_Name, I used "Unknown" because the missing information could not be recovered. For Quantity, Price, and Delivery_Days, I used the mean because the values were relatively balanced and had no major outliers. For Country and Product, I used the mode because they are categorical columns. This assignment helped me understand that data cleaning depends on the type of data and the problem being solved. If 50% of the Price values were missing, I would investigate the reason for the missing data before filling it because using the mean for a large amount of missing data could affect the accuracy of the analysis.
